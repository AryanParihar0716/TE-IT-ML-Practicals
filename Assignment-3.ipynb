```python
#1st Graph
sns.countplot(data=df,x=df["Na"],color="red",native_scale=True)
```

```python
"""Details Understood
    1.Countplot shows the counts of the observations
    2.It gives clear representation about the values scattered across the dataset
    3.native_scale allows us to group similar entires together as it converts them to ordinal values
"""
```

```python
#2nd Graph
sns.barplot(df, x="Mg", y="Fe",palette=["red","green","blue"])
```

```python
"""Details Understood
    1.Barplot shows the  point estimates and errors as rectangular bars.
    2.It gives clear representation about the values scattered across the dataset
    3.Here both axis values must be mentioned to get clear graph(Otherwise it will simply give the only as a huge rectangle)
"""
```

```python
#3rd Graph
sns.stripplot(df, x="Al")
```

```python
"""Details Understood
    1.Stripplot draws a categorical scatterplot using jitter to reduce overplotting.
    2.A strip plot can be drawn on its own, but it is also a good complement to a box or violin plot in cases where you want to show all observations along with some representation of the underlying distribution.
    3.Here both axis values must be mentioned to get clear graph(Otherwise it only produces that data in either horizontal or vertical position)
    """
```

```python
#4th Graph
sns.boxplot(x=df["RI"])
```

```python
"""Details Understood
    1.Boxplot draws a box plot to show distributions with respect to categories.
    2.A box plot (or box-and-whisker plot) shows the distribution of quantitative data in a way that facilitates comparisons between variables or across levels of a categorical variable.
    3.Here both axis values must be mentioned to get clear graph(Otherwise it only produces that data in either horizontal or vertical position)
    """
```

```python
#5th Graph
sns.histplot(data=df, x="Na")
```

```python
"""Details Understood
    1.Histplot is used to plot univariate or bivariate histograms to show distributions of datasets..
    2.A histogram is a classic visualization tool that represents the distribution of one or more variables by counting the number of observations that fall within discrete bins.
    3.Here if only one axis parameter is mentioned it is fine as it uses the other asix to show the plot according to how a stat is used.
    """
```

```python
#6th Graph
sns.scatterplot(data=df,x="RI",y="Type")
```

```python
"""Details Understood
    1.Histplot is used to draw a scatter plot with possibility of several semantic groupings.
    2.Both axis parameters need to be mentioned for accurate representation of the points on the graph
    """
```

```python
#7th Graph
sns.pairplot(df, hue="Type")
```

```python
"""Details Understood
    1.Pairplot is used to plot pairwise relationships in a dataset.
    2.No need of axes parameters."""
```

```python
#8th Graph
sns.rugplot(data=df, x="K", y="Type")
```

```python
"""Details Understood
    1.Rugplot is used to plot marginal distributions by drawing ticks along the x and y axes.
    2.Axes parameters must be defined to get proper visual."""
```
